# Yb⁺: cooling power–detuning heat maps

**비교 대상:** (1) 연속 냉각, (2) 냉각–암시간 반복, (3) 냉각–암시간 중 여기 펄스 반복.

가로축은 부호가 있는 detuning $\delta/2\pi$ [MHz], 세로축은 **공명 포화도 $s$**입니다.
기본 입력은 대화의 참고 조건: $f=1.6$ MHz, cooling/pulse 각도 각각 $40^\circ$, $s=2$, red detuning $-20$ MHz,
냉각 $1\,\mu$s/try, 암시간 $0.58\,\mu$s, 환경 가열 $10$ quanta/ms입니다. MFD는 $50\,\mu$m입니다.
이 값들은 실제 측정값으로 수정할 수 있습니다. 이 notebook은 장비와 통신하거나 GUI 설정을 변경하지 않습니다.

**실행:** 아래 `CFG` 셀을 수정한 뒤 **Restart Kernel and Run All**. 필요한 패키지는 NumPy, Matplotlib, pandas, IPython입니다.
그림과 표는 notebook 안에 포함되며, 마지막 셀의 `SAVE_OUTPUTS=True`로 PNG/CSV/NPZ/설정 JSON을 내보낼 수 있습니다.

기본 모델은 **단일 진행파를 받는 유효 2준위 이온의 optical Bloch force spectrum**입니다.
포화 효과와 유한한 트랩 주파수를 포함하되 운동은 Lamb–Dicke 2차 및 운동 위상 평균 근사입니다.
실제 $^{171}$Yb⁺ hyperfine/Zeeman/repump/dark state, micromotion, 주기적 radiation-pressure 힘에 의한 coherent 구동은 포함하지 않습니다.
따라서 계산의 수치적 검증과 실험 모델의 정확도를 구분해야 합니다. 현재 입력은 **한 모드**를 뜻하며,
세 직교 모드가 한 빔과 모두 $40^\circ$일 수는 없습니다. 각 모드는 자신의 주파수·각도·가열률로 별도 계산합니다.

## 1. 세 경우와 관측 시점

| 경우 | 냉각 | 외부 환경 가열 | 여기 펄스 |
|---|---|---|---|
| 1. Continuous cooling | 항상 ON | 기본 OFF: 이상적인 laser-only 기준. `case1_include_environment=True`로 같은 환경 가열 포함 가능 | 없음 |
| 2. Cooling + dark | $t_c$ ON → $t_d$ OFF 반복 | **ON/OFF 모두** $H$ quanta/s | 없음 |
| 3. Cooling + dark + pulse | case 2와 동일 | case 2와 동일 | 암시간의 $t_p$에 1회 |

case 1의 laser-only는 냉각광 자체의 recoil diffusion까지 제거한다는 뜻이 아닙니다. 그 diffusion은 모든 경우의 $A_+$에 이미 포함됩니다.
외부 가열의 효과만 공정하게 비교하려면 `case1_include_environment=True`를 사용하십시오. 연속 냉각 + 같은 환경의 결과도 별도로 계산합니다.

반복 조건의 정상상태는 일정한 한 값이 아니라 주기적으로 변합니다. 기본 heat map은 **주기 평균**이며,
`n_view`를 `pre_pulse`, `post_pulse`, `after_cooling`, `end_dark`로 바꿀 수 있습니다.
HOM 광자 방출 시 운동 상태를 보려면 `pre_pulse`가 유용합니다.
`pulse_time_dark_us=0.25`는 OFF 이후 펄스 시점의 **예시값**으로, 광학적 지연을 측정해 바꾸십시오.

이 notebook은 사용자 지정의 **1 µs 냉각당 1 try**를 모델링합니다. 기존 HOM의 2 trials/group 전체 시퀀스를 자동으로 읽지 않습니다.
계산 중 AOM ON 시간을 알고 있으면 `extra_cooling_us`에 더할 수 있습니다. 하나의 ON 구간으로 합치는 것은 위상 평균 rate 모델에서의 근사입니다.

```python
import json
import math
from pathlib import Path
from dataclasses import dataclass, asdict, replace

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, Normalize, SymLogNorm
from matplotlib.ticker import LogLocator
from IPython.display import display, Markdown

plt.rcParams.update({
    'figure.dpi': 115, 'savefig.dpi': 180,
    'font.family': 'DejaVu Sans', 'font.size': 10,
    'axes.titlesize': 12, 'axes.labelsize': 11,
    'axes.spines.top': False, 'axes.spines.right': False,
    'mathtext.fontset': 'dejavusans',
})

H_PLANCK = 6.62607015e-34
HBAR = H_PLANCK / (2*np.pi)
K_B = 1.380649e-23
U_ATOMIC = 1.66053906892e-27
C_LIGHT = 299792458.0
FIGURES = {}
```

```python
@dataclass(frozen=True)
class Parameters:
    # One mode of one ion. The mass default is 171Yb (ion/electron correction is negligible here).
    mass_u: float = 170.936323
    wavelength_nm: float = 369.5
    linewidth_MHz: float = 19.6  # Gamma/(2*pi); Gamma is a population decay rate [s^-1].
    mode_MHz: float = 1.6
    cooling_angle_deg: float = 40.0
    pulse_angle_deg: float = 40.0
    emission_xi_cooling: float = 1/3  # <(emission direction . mode axis)^2>
    emission_xi_pulse: float = 1/3

    environment_quanta_per_ms: float = 10.0
    cooling_us: float = 1.0
    extra_cooling_us: float = 0.0
    dark_us: float = 0.58
    pulse_time_dark_us: float = 0.25
    pulse_excitation_probability: float = 1.0
    extra_pulse_quanta: float = 0.0  # Mean extra recoil/repump quanta per try, if calibrated.
    case1_include_environment: bool = False

    s_reference: float = 2.0
    detuning_reference_MHz: float = -20.0  # Red detuning is NEGATIVE.
    s_min: float = 0.01
    s_max: float = 30.0
    s_points: int = 121
    detuning_min_MHz: float = -80.0
    detuning_max_MHz: float = 10.0
    detuning_points: int = 241
    n_view: str = 'cycle_mean'

    # Validity / useful steady-state filters. These are diagnostic choices, not universal bounds.
    ld_diagnostic_limit: float = 0.30
    max_energy_relaxation_ms: float = 1.0
    initial_n_for_transient: float = 30.0
    transient_cycles: int = 600

    # User-supplied MFD; no effect on an s-axis heat map.
    mfd_um: float = 50.0  # ASSUMPTION: Gaussian 1/e^2 intensity DIAMETER at the ion.
    measured_I_sat_W_m2: float | None = None  # Optional effective saturation intensity calibration.

CFG = Parameters()

# Examples (uncomment and then Run All):
# CFG = replace(CFG, cooling_us=2.0, dark_us=0.8, pulse_time_dark_us=0.25)
# CFG = replace(CFG, n_view='pre_pulse', case1_include_environment=True)
# CFG = replace(CFG, mode_MHz=2.2, cooling_angle_deg=55, pulse_angle_deg=30)
# CFG = replace(CFG, mass_u=173.938859)  # 174Yb mass; this alone does NOT implement Zeeman multilevel dynamics.

def validate_parameters(c):
    positive = ('mass_u', 'wavelength_nm', 'linewidth_MHz', 'mode_MHz', 'cooling_us',
                's_min', 's_max', 's_reference', 'mfd_um', 'max_energy_relaxation_ms')
    for key in positive:
        if not np.isfinite(getattr(c, key)) or getattr(c, key) <= 0:
            raise ValueError(f'{key} must be finite and positive')
    for key in ('extra_cooling_us', 'dark_us', 'environment_quanta_per_ms', 'extra_pulse_quanta'):
        if not np.isfinite(getattr(c, key)) or getattr(c, key) < 0:
            raise ValueError(f'{key} must be finite and nonnegative')
    if not 0 <= c.pulse_time_dark_us <= c.dark_us:
        raise ValueError('pulse_time_dark_us must lie in [0, dark_us]')
    if not 0 <= c.pulse_excitation_probability <= 1:
        raise ValueError('pulse_excitation_probability must lie in [0, 1]')
    if any(not 0 <= x <= 1 for x in (c.emission_xi_cooling, c.emission_xi_pulse)):
        raise ValueError('emission xi must lie in [0, 1]')
    if c.s_max <= c.s_min or c.detuning_max_MHz <= c.detuning_min_MHz:
        raise ValueError('Scan bounds must be increasing')
    if c.s_points < 3 or c.detuning_points < 3 or c.transient_cycles < 1:
        raise ValueError('Grid sizes / transient_cycles are too small')
    if c.n_view not in ('cycle_mean', 'pre_pulse', 'post_pulse', 'after_cooling', 'end_dark'):
        raise ValueError('Unknown n_view')
    if c.measured_I_sat_W_m2 is not None and c.measured_I_sat_W_m2 <= 0:
        raise ValueError('measured_I_sat_W_m2 must be positive')

def derived(c):
    validate_parameters(c)
    mass = c.mass_u * U_ATOMIC
    wavelength = c.wavelength_nm * 1e-9
    gamma = 2*np.pi*c.linewidth_MHz*1e6
    omega = 2*np.pi*c.mode_MHz*1e6
    k = 2*np.pi/wavelength
    eta2 = HBAR*k*k/(2*mass*omega)
    tc = (c.cooling_us+c.extra_cooling_us)*1e-6
    td = c.dark_us*1e-6
    q = c.pulse_excitation_probability*eta2*(np.cos(np.deg2rad(c.pulse_angle_deg))**2
         + c.emission_xi_pulse) + c.extra_pulse_quanta
    return dict(mass=mass, wavelength=wavelength, gamma=gamma, omega=omega, k=k,
                eta2=eta2, recoil_Hz=eta2*c.mode_MHz*1e6,
                tc=tc, td=td, tp=c.pulse_time_dark_us*1e-6, period=tc+td,
                duty=tc/(tc+td), background=c.environment_quanta_per_ms*1e3, q_pulse=q)

P = derived(CFG)
display(pd.DataFrame([
    ('Mode frequency', CFG.mode_MHz, 'MHz'),
    ('Cooling / dark', f'{P["tc"]*1e6:g} / {P["td"]*1e6:g}', 'us'),
    ('Try period / repetition rate', f'{P["period"]*1e6:.3f} / {1/P["period"]/1e6:.3f}', 'us / MHz'),
    ('Cooling duty', P['duty'], 'fraction'),
    ('Photon recoil frequency E_R/h', P['recoil_Hz']/1e3, 'kHz'),
    ('Pulse recoil per try (one mode)', P['q_pulse'], 'quanta'),
    ('Environment / pulse input', f'{CFG.environment_quanta_per_ms:.3f} / {P["q_pulse"]/P["period"]/1e3:.3f}', 'quanta/ms'),
], columns=['Quantity', 'Value', 'Unit']))
```

## 2. Power, detuning, recoil: conventions

$$s=\frac{I}{I_{\rm sat}}=\frac{2\Omega^2}{\Gamma^2},\qquad
\delta=\omega_L-\omega_0,\qquad \delta<0\;\text{for red detuning}.$$

$s$ is the **on-resonance** saturation parameter, not $s/[1+(2\delta/\Gamma)^2]$.
All internal angular frequencies $\Gamma,\Omega,\delta,\omega_m$ use rad/s; scan coordinates use ordinary MHz.
For a Gaussian beam with $1/e^2$ intensity radius $w=\mathrm{MFD}/2$,

$$I(0)=\frac{2P_{\rm laser}}{\pi w^2},\qquad
P_{\rm laser}=\frac{\pi w^2 I_{\rm sat}}{2}\,s.$$

An ideal closed 2-level dipole transition has $I_{\rm sat}=\pi hc\Gamma/(3\lambda^3)$.
Real hyperfine/Zeeman transitions require an effective intensity calibration; beam MFD must be the value **at the ion**.
The supplied MFD $50\,\mu$m is used only for the following optional conversion, never to change the heat-map $s$ axis.

For one mode,
$$E_R=\frac{(\hbar k)^2}{2m},\quad x_0=\sqrt{\frac{\hbar}{2m\omega_m}},\quad
\eta_0^2=(kx_0)^2=\frac{E_R}{\hbar\omega_m},\quad
\eta_c=\eta_0\cos\theta_c.$$
The effective mean kick per try is
$$q=p_e\eta_0^2\left(\cos^2\theta_p+\xi_p\right)+q_{\rm extra},\quad
\xi_p=\langle(\hat{\mathbf k}_{\rm em}\cdot\hat{\mathbf e}_m)^2\rangle.$$
Isotropic emission gives $\xi=1/3$. Real dipole patterns can be entered separately for cooling and excitation.
The pulse is assumed short, followed by one spontaneous decay; repump recoil can be entered through $q_{\rm extra}$.
This adds mean phonon number without imposing a thermal distribution. It assumes phase-randomized motional kicks. [R1, R4]

```python
w = CFG.mfd_um*1e-6/2
ideal_I_sat = np.pi*H_PLANCK*C_LIGHT*P['gamma']/(3*P['wavelength']**3)
conversion_I_sat = ideal_I_sat if CFG.measured_I_sat_W_m2 is None else CFG.measured_I_sat_W_m2
power_per_s_uW = np.pi*w*w*conversion_I_sat/2*1e6
print(f'Assumed Gaussian waist at ion: {w*1e6:.2f} um (MFD {CFG.mfd_um:g} um)')
print(f'P per unit s: {power_per_s_uW:.4f} uW; s={CFG.s_reference:g}: {CFG.s_reference*power_per_s_uW:.4f} uW')
print('Power conversion:', 'IDEAL 2-LEVEL estimate, not an experimental calibration'
      if CFG.measured_I_sat_W_m2 is None else 'using user-supplied effective I_sat')
```

## 3. Optical Bloch force-spectrum model

The internal master equation, evaluated at the trap centre, is
$$\mathcal L\rho=-i[H_0/\hbar,\rho]+\Gamma\mathcal D[\sigma_-]\rho,
\qquad H_0/\hbar=-\delta|e\rangle\langle e|+\frac{\Omega}{2}(\sigma_++\sigma_-),$$
with $\mathcal D[L]\rho=L\rho L^\dagger-\{L^\dagger L,\rho\}/2$ and $\mathcal L\rho_{ss}=0$.
The exact internal steady scattering rate of this 2-level model is
$$R_{sc}=\Gamma\rho_{ee}=\frac{\Gamma}{2}\frac{s}{1+s+(2\delta/\Gamma)^2}.$$

To first order in displacement, the running-wave interaction is $H_{\rm int}^{(1)}=\hbar(a+a^\dagger)V$,
where a consistent phase convention gives $V=\eta_c\Omega\sigma_y/2$.
The sign of $V$ does not change its autocorrelation. Subtract its steady mean, $\delta V=V-\langle V\rangle_{ss}$, and define
$$S_V(\nu)=\int_0^\infty e^{i\nu t}\langle\delta V(t)\delta V(0)\rangle_{ss}\,dt.$$
Using quantum regression, the notebook evaluates this as the finite matrix resolvent
$$S_V(\nu)=-\operatorname{Tr}\left[
V(\mathcal L+i\nu)^{-1}\{V\rho_{ss}-\langle V\rangle_{ss}\rho_{ss}\}\right].$$
The mean subtraction removes the elastic/constant-force contribution. The static displacement and coherent force from switching the beam are outside this phase-averaged model.

The phonon transition coefficients are [R2, R3]
$$\boxed{A_-=2\operatorname{Re}S_V(+\omega_m)+D_{sp},\qquad
A_+=2\operatorname{Re}S_V(-\omega_m)+D_{sp}},\qquad
D_{sp}=\xi_c\eta_0^2\Gamma\rho_{ee}.$$
Here $D_{sp}$ is defined as the **full equal contribution to each $A_\pm$**; some references put a factor 2 into the definition of $D$ instead.

Thus [R1]
$$\dot{\bar n}=-\kappa\bar n+A_++H,\qquad
\kappa=A_--A_+,\qquad \bar n_L=A_+/\kappa.$$
This treats internal saturation with optical Bloch equations rather than replacing all absorption events by independent Poisson events.
It remains a **2-level, Lamb–Dicke, secular / weak motional coupling model**, not an exact multilevel Yb⁺ calculation.
If $\kappa\le0$, there is no attracting stationary solution: the stationary heat maps mask that point.

```python
def _batch_kron(a, b):
    shape = np.broadcast_shapes(a.shape[:-2], b.shape[:-2])
    return np.einsum('...ij,...kl->...ikjl', a, b).reshape(shape+(4, 4))

def cooling_rates(s, detuning_MHz, c=CFG):
    """Lamb-Dicke A-/A+ from the exact internal two-level force spectrum.

    Arrays broadcast. Column-major density-matrix vectorization; all matrices
    are scaled by Gamma to keep the linear systems numerically well conditioned.
    Returns rates in s^-1, not angular frequencies or Hz/2pi.
    """
    p = derived(c)
    s, d = np.broadcast_arrays(np.asarray(s, float), np.asarray(detuning_MHz, float))
    if np.any(~np.isfinite(s)) or np.any(s < 0) or np.any(~np.isfinite(d)):
        raise ValueError('s must be nonnegative and scan values finite')
    eye = np.eye(2, dtype=complex)
    sm = np.array([[0, 1], [0, 0]], complex)  # basis (g,e)
    pe = sm.conj().T @ sm
    sx = sm + sm.conj().T
    sy = np.array([[0, -1j], [1j, 0]], complex)
    omega_rabi_over_gamma = np.sqrt(s/2)
    ham = -(d/c.linewidth_MHz)[...,None,None]*pe + omega_rabi_over_gamma[...,None,None]*sx/2
    L = -1j*(_batch_kron(eye, ham)-_batch_kron(ham.swapaxes(-1,-2), eye))
    L += _batch_kron(sm.conj(), sm)-0.5*(_batch_kron(eye, pe)+_batch_kron(pe.T, eye))
    stationary_system = L.copy()
    stationary_system[..., -1, :] = [1, 0, 0, 1]
    rhs = np.zeros(s.shape+(4,), complex)
    rhs[..., -1] = 1
    rho_vec = np.linalg.solve(stationary_system, rhs[...,None])[...,0]
    rho = rho_vec.reshape(s.shape+(2,2)).swapaxes(-1,-2)
    eta_c = np.sqrt(p['eta2'])*np.cos(np.deg2rad(c.cooling_angle_deg))
    V = eta_c*omega_rabi_over_gamma[...,None,None]*sy/2
    mean_V = np.trace(V@rho, axis1=-2, axis2=-1)
    source = V@rho-mean_V[...,None,None]*rho
    source_vec = source.swapaxes(-1,-2).reshape(s.shape+(4,))

    def real_spectrum(nu_over_gamma):
        response_vec = -np.linalg.solve(L+1j*nu_over_gamma*np.eye(4), source_vec[...,None])[...,0]
        response = response_vec.reshape(s.shape+(2,2)).swapaxes(-1,-2)
        return np.trace(V@response, axis1=-2, axis2=-1).real

    rho_ee = rho[...,1,1].real
    emission = c.emission_xi_cooling*p['eta2']*rho_ee
    A_minus = p['gamma']*(2*real_spectrum(+p['omega']/p['gamma'])+emission)
    A_plus = p['gamma']*(2*real_spectrum(-p['omega']/p['gamma'])+emission)
    if np.any(A_minus < -1e-8) or np.any(A_plus < -1e-8):
        raise ArithmeticError('Unexpected negative quantum-noise spectrum')
    A_minus, A_plus = np.maximum(A_minus, 0), np.maximum(A_plus, 0)
    kappa = A_minus-A_plus
    stable = kappa > 1e-9
    n_laser = np.divide(A_plus, kappa, out=np.full(s.shape, np.nan), where=stable)
    return dict(A_minus=A_minus, A_plus=A_plus, kappa=kappa, n_laser=n_laser,
                scattering=p['gamma']*rho_ee, stable=stable, rho_ee=rho_ee)

def simple_doppler(s, detuning_MHz, c=CFG):
    """Diagnostic semiclassical friction + independent-scattering approximation.

    This is the earlier back-of-envelope model, not the default heat-map engine.
    n = E/(hbar*omega)-1/2: do not silently omit the zero-point offset.
    """
    p = derived(c)
    s, d = np.broadcast_arrays(np.asarray(s, float), np.asarray(detuning_MHz, float))
    z = d/c.linewidth_MHz
    den = 1+s+4*z*z
    mu2 = np.cos(np.deg2rad(c.cooling_angle_deg))**2
    rate = p['gamma']*s/(2*den)
    kappa = -4*HBAR*p['k']**2/p['mass']*mu2*s*z/(den*den)
    diffusion_quanta = rate*p['eta2']*(mu2+c.emission_xi_cooling)
    n = np.divide(diffusion_quanta, kappa, out=np.full(s.shape, np.nan), where=kappa>0)-0.5
    return dict(kappa=kappa, n_laser=n, scattering=rate)
```

## 4. Exact periodic solution of the mean-phonon rate equation

The following closed forms are derived here by composing the ON and OFF solutions; **exact** refers to the stated constant-rate model.
Let $T=t_c+t_d$, $a=e^{-\kappa t_c}$, $H$ be environmental heating in quanta/s, and
$$n_c=\frac{A_++H}{\kappa}.$$
Environmental heating acts during cooling too. Define $n_H$ at the **end of the dark interval** and $n_C$ immediately after cooling:
$$n_{H,k+1}=n_c+a(n_{H,k}-n_c)+Ht_d+q.$$
For case 2 set $q=0$; for case 3 use the pulse recoil. The fixed point is
$$\boxed{n_H=n_c+\frac{Ht_d+q}{1-a},\qquad
n_C=n_c+\frac{a(Ht_d+q)}{1-a}}.$$
At the pulse time $0\le t_p\le t_d$,
$$n_{\rm pre}=n_C+Ht_p,\qquad n_{\rm post}=n_{\rm pre}+q.$$
The cycle mean is
$$\boxed{\langle n\rangle_T=\frac{n_ct_c+(n_H-n_c)(1-a)/\kappa
+n_Ct_d+Ht_d^2/2+q(t_d-t_p)}{T}}.$$
Case 1 is $n_1=A_+/\kappa$ by default, or $(A_++H)/\kappa$ when its environment switch is enabled.
The exact finite-cycle approach is $n_{H,N}=n_H+a^N(n_{H,0}-n_H)$.

**Damping quantities are different:**

* $\kappa$ [s$^{-1}$] is the ON **energy / mean-phonon** relaxation rate. The coherent amplitude decays at $\kappa/2$.
* Per comparison interval $T$, energy retention $r_E=e^{-\kappa T}$ (case 1), or $r_E=e^{-\kappa t_c}$ (cases 2,3).
* Energy damping fraction is $1-r_E$; amplitude retention is $\sqrt{r_E}$.
* Effective energy rate is $\kappa_{eff}=-\ln(r_E)/T$: $\kappa$ for case 1, $D\kappa$ for cases 2,3, $D=t_c/T$.
* The equivalent weak-damping oscillator ratio is
  $$\boxed{\zeta_{eff}=\frac{\kappa_{eff}}{2\omega_m}},\quad
  \ddot x+2\zeta_{eff}\omega_m\dot x+\omega_m^2x=0.$$

This $\zeta_{eff}$ is an equivalent phase-averaged damping ratio, not a Floquet calculation of the RF trap.
**Cases 2 and 3 have identical damping maps in this model.** An additive random recoil changes the fixed point but not the contraction eigenvalue.
It would change the damping only through physics outside this model (e.g. state-dependent cooling or temperature-dependent friction).

```python
CASE_NAMES = ['1 | Continuous cooling', '2 | Cooling + dark', '3 | Cooling + dark + pulse']
VIEW_NAMES = {'cycle_mean': 'Cycle-mean phonon number', 'pre_pulse': 'Phonon number just before pulse',
              'post_pulse': 'Phonon number just after pulse', 'after_cooling': 'Phonon number after cooling',
              'end_dark': 'Phonon number at end of dark interval'}

def three_cases(rates, c=CFG):
    p = derived(c)
    valid = rates['stable']
    kap = np.where(valid, rates['kappa'], np.nan)
    laser = rates['n_laser']
    Henv, tc, td, tp, T = p['background'], p['tc'], p['td'], p['tp'], p['period']
    one_minus_a = -np.expm1(-kap*tc)  # Accurate even for very weak damping.
    a = np.exp(-kap*tc)
    n_cont_bg = laser+Henv/kap
    n1 = n_cont_bg if c.case1_include_environment else laser

    def metadata(data, k_eff):
        data.update(kappa_eff=k_eff, zeta_eff=k_eff/(2*p['omega']),
                    energy_retention=np.exp(-k_eff*T),
                    damping_fraction=-np.expm1(-k_eff*T), tau_ms=1e3/k_eff)
        # Conservative diagnostic uses the largest occupancy within the cycle and full k.
        data['ld_parameter'] = p['eta2']*(2*np.maximum(data['end_dark'], data['after_cooling'])+1)
        return data

    result = [metadata({name: n1 for name in VIEW_NAMES}, kap)]
    for q in (0.0, p['q_pulse']):
        gain = Henv*td+q
        hi = n_cont_bg+gain/one_minus_a
        lo = n_cont_bg+a*gain/one_minus_a
        pre = lo+Henv*tp
        integral_on = n_cont_bg*tc+(hi-n_cont_bg)*one_minus_a/kap
        integral_off = lo*td+0.5*Henv*td*td+q*(td-tp)
        result.append(metadata(dict(end_dark=hi, after_cooling=lo, pre_pulse=pre,
                                    post_pulse=pre+q, cycle_mean=(integral_on+integral_off)/T),
                               p['duty']*kap))
    return result

# Keep reference values on the grid; axes remain increasing and the default logarithmic s grid is preserved.
s_values = np.unique(np.r_[np.geomspace(CFG.s_min, CFG.s_max, CFG.s_points),
                          [CFG.s_reference] if CFG.s_min <= CFG.s_reference <= CFG.s_max else []])
d_values = np.unique(np.r_[np.linspace(CFG.detuning_min_MHz, CFG.detuning_max_MHz, CFG.detuning_points),
                          [CFG.detuning_reference_MHz] if CFG.detuning_min_MHz <= CFG.detuning_reference_MHz <= CFG.detuning_max_MHz else []])
DETUNING, SATURATION = np.meshgrid(d_values, s_values)
RATES = cooling_rates(SATURATION, DETUNING)
CASES = three_cases(RATES)
REF_RATES = cooling_rates(CFG.s_reference, CFG.detuning_reference_MHz)
REF_CASES = three_cases(REF_RATES)
print(f'Computed {SATURATION.size:,} points; {np.count_nonzero(RATES["stable"]):,} have positive damping.')

summary_rows = []
for name, case in zip(CASE_NAMES, REF_CASES):
    summary_rows.append(dict(case=name, n_mean=float(case['cycle_mean']), n_pre=float(case['pre_pulse']),
                             n_post=float(case['post_pulse']), n_after_cooling=float(case['after_cooling']),
                             n_end_dark=float(case['end_dark']), kappa_eff_per_s=float(case['kappa_eff']),
                             tau_energy_us=float(case['tau_ms']*1000), zeta_eff=float(case['zeta_eff']),
                             damping_per_T_percent=float(case['damping_fraction']*100)))
REFERENCE_TABLE = pd.DataFrame(summary_rows)
display(REFERENCE_TABLE.round(6))
print(f'Reference ON scattering: {float(REF_RATES["scattering"])/1e6:.4f} million photons/s')
if bool(REF_RATES['stable']):
    print(f'Reference ON cooling time: {1e6/float(REF_RATES["kappa"]):.3f} us')
else:
    print('Reference has no attracting stationary solution (kappa <= 0).')
```

## 5. Numerical checks before plotting

These checks test independent limits and exact identities, rather than comparing a function to a copy of itself:

1. Internal steady population agrees with the analytic 2-level scattering rate.
2. In the low-saturation limit,
   $A_-=\eta_c^2W(\delta+\omega_m)+\xi_c\eta_0^2W(\delta)$ and
   $A_+=\eta_c^2W(\delta-\omega_m)+\xi_c\eta_0^2W(\delta)$, with
   $W(\delta)=\Gamma s/[2(1+4\delta^2/\Gamma^2)]$. [R1]
3. At small trap frequency the QRT damping agrees with the derivative of the radiation-pressure force.
4. Cases coincide when added heating and recoil vanish; zero dark time gives the continuous result.
5. Repeated affine propagation agrees with the analytic fixed point, and a sampled trajectory reproduces the cycle mean.

```python
def run_checks():
    # Independent standard fixtures: changing the plotted geometry must not invalidate the tests.
    c = Parameters()
    p = derived(c)
    checked = []
    ss = np.array([0.001, 0.1, 2.0, 20.0])
    dd = np.array([-60., -20., -5., 5.])
    rr = cooling_rates(ss, dd, c)
    analytic_scattering = p['gamma']/2*ss/(1+ss+4*(dd/c.linewidth_MHz)**2)
    np.testing.assert_allclose(rr['scattering'], analytic_scattering, rtol=1e-12)
    checked.append('OBE stationary population vs analytic scattering')

    weak = 1e-7
    dd = np.array([-40., -20., -5.])
    rr = cooling_rates(weak, dd, c)
    mu2 = np.cos(np.deg2rad(c.cooling_angle_deg))**2
    def W(det):
        return p['gamma']/2*weak/(1+4*(det/c.linewidth_MHz)**2)
    low_minus = p['eta2']*(mu2*W(dd+c.mode_MHz)+c.emission_xi_cooling*W(dd))
    low_plus = p['eta2']*(mu2*W(dd-c.mode_MHz)+c.emission_xi_cooling*W(dd))
    np.testing.assert_allclose(rr['A_minus'], low_minus, rtol=3e-6, atol=1e-13)
    np.testing.assert_allclose(rr['A_plus'], low_plus, rtol=3e-6, atol=1e-13)
    checked.append('Weak-drive resolved-sideband limit (signs and factors of 2)')

    slow_c = replace(c, mode_MHz=0.001)
    kk = cooling_rates(2.0, -20.0, slow_c)['kappa']
    k_classical = simple_doppler(2.0, -20.0, slow_c)['kappa']
    np.testing.assert_allclose(kk, k_classical, rtol=3e-5, atol=1e-7)
    checked.append('Small-mode-frequency friction limit')

    signs = cooling_rates(2., np.array([-20., 0., 20.]), c)['kappa']
    assert signs[0] > 0 and abs(signs[1]) < 1e-6 and signs[2] < 0
    assert not cooling_rates(0., -20., c)['stable']
    checked.append('Red cooling / resonance / blue anti-damping / zero power')

    zc = replace(c, environment_quanta_per_ms=0, pulse_excitation_probability=0,
                 extra_pulse_quanta=0, case1_include_environment=False)
    zcases = three_cases(cooling_rates(2., -20., zc), zc)
    np.testing.assert_allclose([x['cycle_mean'] for x in zcases], zcases[0]['cycle_mean'], rtol=1e-12)
    dc = replace(c, dark_us=0, pulse_time_dark_us=0, pulse_excitation_probability=0,
                 extra_pulse_quanta=0, case1_include_environment=True)
    dcases = three_cases(cooling_rates(2., -20., dc), dc)
    np.testing.assert_allclose([x['cycle_mean'] for x in dcases], dcases[0]['cycle_mean'], rtol=1e-12)
    checked.append('No-added-heating and zero-dark-time limits')

    # Use a fixed stable test point even if the user's marker is blue-detuned.
    test_r = cooling_rates(2., -20., c)
    test_cases = three_cases(test_r, c)
    kap = float(test_r['kappa'])
    a = np.exp(-kap*p['tc'])
    nc = (float(test_r['A_plus'])+p['background'])/kap
    q, Henv, tc, td, tp = p['q_pulse'], p['background'], p['tc'], p['td'], p['tp']
    hi = float(test_cases[2]['end_dark'])
    value = hi+3.0
    for _ in range(1000):
        value = nc+(value-nc)*a+Henv*td+q
    expected = hi+3*np.exp(-1000*kap*tc)
    np.testing.assert_allclose(value, expected, rtol=2e-11, atol=1e-10)
    np.testing.assert_allclose(test_cases[2]['end_dark']-test_cases[1]['end_dark'],
                               q/(-np.expm1(-kap*tc)), rtol=1e-10, atol=1e-11)
    np.testing.assert_allclose(test_cases[1]['zeta_eff'], test_cases[2]['zeta_eff'])
    checked.append('Fixed point, transient contraction, additive pulse increment')

    u = np.linspace(0, tc, 20001)
    on = nc+(hi-nc)*np.exp(-kap*u)
    lo = on[-1]
    v1, v2 = np.linspace(0, tp, 10001), np.linspace(tp, td, 10001)
    integral = np.trapz(on, u)+np.trapz(lo+Henv*v1, v1)+np.trapz(lo+Henv*v2+q, v2)
    np.testing.assert_allclose(integral/p['period'], test_cases[2]['cycle_mean'], rtol=2e-9)
    checked.append('Piecewise trajectory quadrature vs cycle mean')
    print('PASS: ' + '\nPASS: '.join(checked))
    return checked

CHECKS_PASSED = run_checks()
```

## 6. Main heat maps: final phonon number and damping ratio

Columns share a colour scale within each row. Grey = no attracting stationary solution ($\kappa\le0$).
White stars mark the reference input; a white dashed contour, when present, marks the chosen Lamb–Dicke diagnostic boundary
$\eta_0^2(2n_{\max}+1)=\texttt{ld_diagnostic_limit}$.
Points outside this boundary are still displayed as model extrapolations; do not treat them as precise predictions.
The diagnostic uses full $k$ conservatively, not only the cooling-beam projection.

```python
def positive_norm(arrays):
    vals = np.concatenate([np.asarray(x)[np.isfinite(x) & (np.asarray(x)>0)].ravel() for x in arrays])
    if vals.size == 0:
        return LogNorm(1, 10)
    lo, hi = float(vals.min()), float(vals.max())
    return LogNorm(max(lo, 1e-12), max(hi, lo*1.01))

def configure_map_axis(ax):
    ax.set_yscale('log')
    ax.set_xlabel(r'Detuning $\delta/2\pi$ (MHz)')
    ax.set_ylabel(r'On-resonance saturation $s$')
    ax.set_xlim(d_values[0], d_values[-1])
    ax.set_ylim(s_values[0], s_values[-1])
    ax.set_facecolor('#e5e7eb')
    if d_values[0] <= 0 <= d_values[-1]:
        ax.axvline(0, color='#566174', lw=0.8, ls=':')
    if d_values[0] <= CFG.detuning_reference_MHz <= d_values[-1] and s_values[0] <= CFG.s_reference <= s_values[-1]:
        ax.plot(CFG.detuning_reference_MHz, CFG.s_reference, marker='*', ms=11,
                mfc='white', mec='#172033', mew=0.8, zorder=5)

def heatmap(ax, values, norm, cmap='viridis', validity=None):
    cmap_obj = plt.get_cmap(cmap).copy()
    cmap_obj.set_bad('#e5e7eb')
    im = ax.pcolormesh(d_values, s_values, np.ma.masked_invalid(values),
                       norm=norm, cmap=cmap_obj, shading='auto', rasterized=True)
    configure_map_axis(ax)
    if validity is not None:
        finite = validity[np.isfinite(validity)]
        level = CFG.ld_diagnostic_limit
        if finite.size and finite.min() < level < finite.max():
            ax.contour(d_values, s_values, np.ma.masked_invalid(validity), levels=[level],
                       colors=['white'], linestyles=['--'], linewidths=[0.9])
    return im

fig, axes = plt.subplots(2, 3, figsize=(16, 9), constrained_layout=True)
n_norm = positive_norm([x[CFG.n_view] for x in CASES])
z_norm = positive_norm([x['zeta_eff'] for x in CASES])
for j, (name, result) in enumerate(zip(CASE_NAMES, CASES)):
    im_n = heatmap(axes[0,j], result[CFG.n_view], n_norm, validity=result['ld_parameter'])
    im_z = heatmap(axes[1,j], result['zeta_eff'], z_norm, cmap='magma')
    axes[0,j].set_title(name)
    axes[1,j].set_title(name)
fig.colorbar(im_n, ax=axes[0,:], label=VIEW_NAMES[CFG.n_view]+r' $\bar n$', shrink=0.94)
fig.colorbar(im_z, ax=axes[1,:], label=r'Equivalent damping ratio $\zeta_{eff}=\kappa_{eff}/(2\omega_m)$', shrink=0.94)
fig.suptitle('Cooling power and detuning | one mode, effective two-level model', fontsize=17)
FIGURES['01_phonon_and_damping_heatmaps'] = fig
plt.show()
```

## 8. Extra-heating budget and useful grid minima

The first panel below subtracts **continuous cooling with the same environmental heating** from case 2.
It isolates the penalty from interrupting the cooling. The second panel isolates the added excitation pulse.
The last panel shows the Lamb–Dicke diagnostic for case 3; large values require a model beyond this approximation.
These panels use a logarithmic colour scale above 0.01 (linear below it, so zero remains representable).

The reported minima are **sampled-grid minima**, not guarantees of global optimality.
The eligibility filter requires positive damping, the selected LD diagnostic, and energy relaxation time below
`max_energy_relaxation_ms`. Slow points can look attractive in a steady-state map while being impractical in a finite experiment.

```python
safe_kappa = np.where(RATES['stable'], RATES['kappa'], np.nan)
CONTINUOUS_WITH_ENV = RATES['n_laser']+P['background']/safe_kappa
interrupt_penalty = CASES[1][CFG.n_view]-CONTINUOUS_WITH_ENV
pulse_penalty = CASES[2][CFG.n_view]-CASES[1][CFG.n_view]
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5), constrained_layout=True)
for ax, values, title, label in zip(axes,
        [interrupt_penalty, pulse_penalty, CASES[2]['ld_parameter']],
        ['Penalty from cooling interruptions', 'Penalty from excitation pulse', 'Lamb-Dicke diagnostic | case 3'],
        [r'Added $\bar n$', r'Added $\bar n$', r'$\eta_0^2(2n_{max}+1)$']):
    values = np.where(np.abs(values)<1e-13, 0, values)
    finite = values[np.isfinite(values)]
    vmax = max(float(finite.max()) if finite.size else 1, 1e-10)
    im = heatmap(ax, values, SymLogNorm(linthresh=0.01, vmin=0, vmax=vmax, base=10), cmap='viridis')
    ax.set_title(title)
    fig.colorbar(im, ax=ax, label=label)
FIGURES['03_heating_budget_and_validity'] = fig
plt.show()

minima = []
for name, case in zip(CASE_NAMES, CASES):
    eligible = (RATES['stable'] & np.isfinite(case[CFG.n_view])
                & (case['ld_parameter'] <= CFG.ld_diagnostic_limit)
                & (case['tau_ms'] <= CFG.max_energy_relaxation_ms))
    if not np.any(eligible):
        minima.append(dict(case=name, status='No eligible grid point'))
        continue
    ii, jj = np.unravel_index(np.argmin(np.where(eligible, case[CFG.n_view], np.inf)), SATURATION.shape)
    minima.append(dict(case=name, status='grid minimum', s=SATURATION[ii,jj],
                       detuning_MHz=DETUNING[ii,jj], n=case[CFG.n_view][ii,jj],
                       tau_ms=case['tau_ms'][ii,jj], zeta=case['zeta_eff'][ii,jj],
                       scan_edge=ii in (0,len(s_values)-1) or jj in (0,len(d_values)-1)))
MINIMA_TABLE = pd.DataFrame(minima)
display(MINIMA_TABLE)
```

## 9. Cuts, finite-cycle convergence, and the waveform within a cycle

Steady state must be sampled at a specified phase. Each case's transient below is plotted at the end of each comparison interval.
The initial value is `initial_n_for_transient`; all three cases use the same initial mean.
The jump at the pulse is the absorption + spontaneous-emission mean recoil, compressed into one event on this µs scale.
The within-cycle panel subtracts each case's cycle mean so that its small oscillation and recoil jump remain visible;
the absolute values are in the reference table and the heat maps.

```python
def reference_cycle_trace(case_number, c=CFG, points=300):
    p = derived(c)
    rr = cooling_rates(c.s_reference, c.detuning_reference_MHz, c)
    if not bool(rr['stable']):
        return np.array([]), np.array([])
    cases = three_cases(rr, c)
    case = cases[case_number]
    if case_number == 0:
        return np.array([0, p['period']])*1e6, np.repeat(float(case['cycle_mean']), 2)
    q = p['q_pulse'] if case_number == 2 else 0
    kap = float(rr['kappa'])
    nc = (float(rr['A_plus'])+p['background'])/kap
    u = np.linspace(0, p['tc'], points)
    on = nc+(float(case['end_dark'])-nc)*np.exp(-kap*u)
    before = np.linspace(0, p['tp'], points)
    after = np.linspace(p['tp'], p['td'], points)
    lo = float(case['after_cooling'])
    times = np.r_[u, p['tc']+before, p['tc']+after]*1e6
    values = np.r_[on, lo+p['background']*before, lo+p['background']*after+q]
    return times, values

cut_d = three_cases(cooling_rates(CFG.s_reference, d_values))
cut_s = three_cases(cooling_rates(s_values, CFG.detuning_reference_MHz))
fig, axes = plt.subplots(2, 2, figsize=(13, 9), constrained_layout=True)
colors = ['#2166ac', '#d98224', '#8e3b88']
for j, name in enumerate(CASE_NAMES):
    axes[0,0].plot(d_values, cut_d[j][CFG.n_view], color=colors[j], label=name)
    axes[0,1].plot(s_values, cut_s[j][CFG.n_view], color=colors[j], label=name)
axes[0,0].set(xlabel='Detuning (MHz)', ylabel=VIEW_NAMES[CFG.n_view],
              title=f'Detuning cut at s={CFG.s_reference:g}', yscale='log')
axes[0,1].set(xlabel='On-resonance saturation s', ylabel=VIEW_NAMES[CFG.n_view],
              title=f'Power cut at detuning={CFG.detuning_reference_MHz:g} MHz', xscale='log', yscale='log')

if bool(REF_RATES['stable']):
    periods = np.arange(CFG.transient_cycles+1)
    for j, case in enumerate(REF_CASES):
        target = float(case['end_dark'])
        n_at_boundary = target+(CFG.initial_n_for_transient-target)*np.exp(-float(case['kappa_eff'])*P['period']*periods)
        axes[1,0].plot(periods*P['period']*1e3, n_at_boundary, color=colors[j], label=CASE_NAMES[j])
        t, n = reference_cycle_trace(j)
        axes[1,1].plot(t, n-float(case['cycle_mean']), color=colors[j], label=CASE_NAMES[j])
    axes[1,1].axvline(P['tc']*1e6, color='gray', ls=':', label='Cooling OFF')
    axes[1,1].axvline((P['tc']+P['tp'])*1e6, color='#aa3a3a', ls=':', label='Excitation pulse')
else:
    for ax in axes[1,:]:
        ax.text(.5,.5,'Reference has no stable stationary solution', ha='center', va='center', transform=ax.transAxes)
axes[1,0].set(xlabel='Elapsed time (ms)', ylabel='Phonon number at interval boundary', title='Approach to periodic steady state')
axes[1,1].set(xlabel='Time within cycle (us)', ylabel=r'$\bar n(t)-\langle\bar n\rangle_T$', title='Steady cycle: deviation from each cycle mean')
for ax in axes.flat:
    ax.grid(alpha=.18)
    ax.legend(fontsize=8, loc='best')
FIGURES['04_cuts_and_periodic_dynamics'] = fig
plt.show()
```

## 10. Compare to the earlier simple Doppler estimate

For a running wave, linearizing the mean force $F=\hbar k_c R_{sc}(\delta-k_cv)$ gives
$$\kappa_{\rm sc}=\frac{\hbar k^2\cos^2\theta_c}{m}\frac{dR_{sc}}{d\delta}
=-\frac{4\hbar k^2\cos^2\theta_c}{m}
\frac{s(\delta/\Gamma)}{[1+s+(2\delta/\Gamma)^2]^2}.$$
Treating scattering events as independent gives the approximate heating input
$$B_{\rm sc}=R_{sc}\eta_0^2(\cos^2\theta_c+\xi_c),\qquad
n_{L,\rm sc}=B_{\rm sc}/\kappa_{\rm sc}-\tfrac12.$$
The $-1/2$ converts the classical mean oscillator energy to phonon number. This shortcut is useful for checking scales,
but should not be trusted at arbitrary saturation, near the ground state, or outside weak binding.
The default QRT model does not need this subtraction; its $A_+$ already accounts for quantum sideband asymmetry.
The earlier chat estimate near $\bar n\approx10$ used the shortcut. The model comparison below makes the difference explicit.

```python
SIMPLE = simple_doppler(SATURATION, DETUNING)
ref_simple = simple_doppler(CFG.s_reference, CFG.detuning_reference_MHz)
display(pd.DataFrame([
    dict(model='OBE force spectrum (default)', n_laser=float(REF_RATES['n_laser']),
         kappa_on_per_s=float(REF_RATES['kappa']), scattering_MHz=float(REF_RATES['scattering'])/1e6),
    dict(model='Simple semiclassical estimate', n_laser=float(ref_simple['n_laser']),
         kappa_on_per_s=float(ref_simple['kappa']), scattering_MHz=float(ref_simple['scattering'])/1e6),
]))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)
with np.errstate(divide='ignore', invalid='ignore'):
    diffs = [100*(SIMPLE['n_laser']/RATES['n_laser']-1),
             100*(SIMPLE['kappa']/RATES['kappa']-1)]
for ax, vals, title in zip(axes, diffs, ['Simple vs OBE: laser equilibrium n', 'Simple vs OBE: damping rate']):
    vals = np.where(RATES['stable'], vals, np.nan)
    finite = vals[np.isfinite(vals)]
    lim = max(float(np.max(np.abs(finite))) if finite.size else 1, 1e-4)
    im = heatmap(ax, vals, Normalize(-lim, lim), cmap='coolwarm')
    ax.set_title(title)
    fig.colorbar(im, ax=ax, label='Relative difference (%)')
FIGURES['05_model_comparison'] = fig
plt.show()
```

## 11. Interpretation and accuracy limits

* **Low s:** cooling can be very slow. An infinite-time low-temperature point is not automatically useful in the actual experiment.
* **Large s:** optical saturation is included in the internal 2-level model, but a growing motional spread can invalidate the LD expansion.
* **Blue / zero detuning:** no stable stationary solution in this running-wave model; grey pixels are intentional, not missing calculations.
* **Environmental heating:** the input 10 quanta/ms applies to this one mode and is assumed constant with laser power, detuning and temperature. It is not turned off during cooling.
* **Pulse vs cooling power:** the scan changes cooling-light $s$. The excitation pulse has separately fixed $p_e$; varying the cooling power does not automatically change the π-pulse excitation probability.
* **Damping:** extra additive heating changes $n_{ss}$, not the linear relaxation eigenvalue. Cases 2 and 3 should have equal damping maps.
* **True $^{171}$Yb⁺:** a mass choice does not implement hyperfine physics. Polarization, magnetic field, repump sidebands, optical pumping and dark-state dynamics can alter both scattering and cooling rates. [R5]
* **Timing:** the ON segment is treated as having stationary internal populations. Optical transients, AOM optical response and finite excited-state lifetime at the boundary are not explicitly propagated. For a 2-level ion the natural lifetime is about 8 ns, but multilevel preparation can be much slower.
* **Randomness:** emission direction randomness alone does not randomize the motional phase of a periodic pulse train. Deterministic phase locking, switching the radiation-pressure equilibrium, RF micromotion and mode coupling need a phase-space / Floquet / trajectory treatment. [R4, R6]
* **Distribution:** the calculation predicts mean occupancy, not the full $P(n)$ or a proven thermal state. An equivalent temperature from $\bar n$ is not a thermometry result.

For experiment-grade calibration, measure the cooling time constant and continuous-cooling occupancy versus $(s,\delta)$,
then use those measured $\kappa$ and $A_+=\kappa n_L$ in `three_cases`.
If the measured continuous occupancy already includes environmental heating, first remove $H/\kappa$ to obtain the laser-only $n_L$;
otherwise the environmental contribution would be counted twice.
The periodic formulas and damping definitions remain usable with calibrated rates.

**[R2]** J. I. Cirac, R. Blatt, P. Zoller, and W. D. Phillips, *Laser cooling of trapped ions in a standing wave*, Phys. Rev. A **46**, 2668–2681 (1992).
[DOI](https://doi.org/10.1103/PhysRevA.46.2668).
The paper treats both running and standing waves and derives the motional master equation by expansion about internal steady state.
Used here as the theoretical basis for the 2-level force-spectrum method, not as a claim that this experiment is a standing wave.

**[R3]** C. Wunderlich, G. Morigi, and D. Reiß, *Simultaneous cooling of axial vibrational modes in a linear ion trap*, Phys. Rev. A **72**, 023421 (2005).
[DOI](https://doi.org/10.1103/PhysRevA.72.023421), [author-hosted full text](https://quantenoptik.physik.uni-siegen.de/wp-content/uploads/sites/7/2024/05/QIS16.pdf).
Appendix A, Eqs. (A15)–(A18): force spectra, spontaneous-emission diffusion, and quantum regression.
The numerical Liouvillian here is explicitly specialized to two levels; it is not that paper's multilevel implementation.

**[R4]** A. J. Rasmusson et al., *Measurement-induced heating of a trapped ion*, J. Phys. B **57**, 225002 (2024).
[DOI](https://doi.org/10.1088/1361-6455/ad838a), [author-hosted full text](https://iontrap.physics.indiana.edu/papers/rasmusson2024.pdf).
Connects continuous ambient heating and discrete photon-recoil kicks, including mode geometry, and discusses nonthermal motional distributions.

**[R5]** S. Crain et al., *High-speed low-crosstalk detection of a $^{171}$Yb⁺ qubit using superconducting nanowire single photon detectors*, Commun. Phys. **2**, 97 (2019).
[DOI](https://doi.org/10.1038/s42005-019-0195-8).
The $^{171}$Yb⁺ scattering-rate section (Eq. 6) illustrates why the bright hyperfine manifold differs from a closed two-level atom; also gives $\Gamma=2\pi\times19.6$ MHz.

**[R6]** J. H. Wesenberg et al., *Fluorescence during Doppler cooling of a single trapped atom*, Phys. Rev. A **76**, 053416 (2007).
[DOI](https://doi.org/10.1103/PhysRevA.76.053416), [NIST full text](https://tf.nist.gov/general/pdf/2246.pdf).
Cooling transients, energy-dependent Doppler cooling and effects relevant to recooling-based calibration.
